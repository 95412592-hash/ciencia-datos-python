```
import pandas as pd
datos = { 
            "estudiante": ["Ana","Juan","Carla","Pedro","Lucia"],
            "nota": [8,6,9,5,7]
        }
import pandas as pd
df = pd.DataFrame(datos)
print(df) 

promedio = df["nota"].mean()
print("Promedio de las notas:", promedio)

maxima = df["nota"].max()
print("Nota máxima:", maxima)

minima = df["nota"].min()
print("Nota minima:", minima)    

print("Promedio de las notas:", promedio)
```
